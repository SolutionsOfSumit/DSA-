```python
"""
create a pattern like 

12344
1234
123
1

"""

def execution(n):
    for i in range(n, 0, -1):  #in first loop i = n 
        for j in range(1, i + 1, 1):
            print(j, end=" ")
        print()

# case1, case2 = input("case1 : ") , input("case2 : ")

case1, case2 = 4, 2

print(" \n \n Solution case 1 \n \n")
execution(int(case1))

print(" \n \n Solution case 2 \n \n")
execution(int(case2))
```

```python
# getting the sting of numbers 
n = 4
k = ""
for i in range(1,n+1,1):
    k += f"{i}"

j = 1

while j <= n:
    print(k)
    k = k[:-1]
    j += 1
```

```python

%%time
# Create a pattern if n = 5
"""
    *
   ***
  *****
 *******
*********

"""

def execution(n):
    for i in range(1,n + 1,1):
        print(" "*(n-i), end="")
        print("*"*(i), end="")
        print("*"*(i-1))




case1, case2 = 4, 2
# case1, case2 = input("case1 : ") , input("case2 : ")

print(" \n \n Solution case 1 \n \n")
execution(int(case1))

print(" \n \n Solution case 2 \n \n")
execution(int(case2))
```

```python
%%time
"""
for n = 5

*********
 *******
  *****
   ***
    *

"""

def execution(n):
    s = 0
    for i in range(n, 0, -1):
        print(" "*s, end="")
        print("*"*i, end="")
        print("*"*(i-1))
        s += 1

# case1, case2 = input("case1 : ") , input("case2 : ")

case1, case2 = 4, 2

print(" \n \n Solution case 1 \n \n")
execution(int(case1))

print(" \n \n Solution case 2 \n \n")
execution(int(case2))
```
