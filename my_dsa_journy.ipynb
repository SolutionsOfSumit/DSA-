## 217. Contains Duplicates

for this we can have multiple solutions one  of which is brute force in which we will check every number by matching by every number in the array.

```python
n = [1, 3, 2, 1]
found = False

if not found:
    for i in range(len(n)):
        for j in range(i, (len(n))):
            if i != j:
                if n[i] == n[j]:
                    print(True)
                    found = True
                    break

if not found:
    print(False)

```

This proposed solution had a time complexity of O(n*2) The question is if i can solve this by reducing the time complexity to O(n)

```python

n = [1,2,3,1]

spn = set()
found = False

if not found:
    for i in n: 
        if i not in spn:
            spn.add(i)
        else:
            print(True)
            found = True
            break

if not found:
    print(False)


```

## 1. Sum of two numbers

You will be given an array and a target you have to return indices of the number which will return in the target. There is only one solution.

### Attempt 1 Self

```python
array = [ 2, 7, 11, 15]

target = 9

for i in range(len(array)):
    for j in range(len(array)):
        if i != j: 
            if array[i] + array[j] == target:
                print(i,j)
```

This had time complexity of O(n*2) now we have to solve it for O(n)
Hint = Using hash maps
### Attempt 2
Theory if the target is certain and we can have only 2 solution so we can say target - array[i] = something now we have to find that something in the array if it exist then we have our solution.

```python
array = [ 2, 7, 11, 15]

target = 9
hm = {}

for i in range(len(array)):
    ans = target - array[i]
    if ans in hm.keys():
        print(i, hm[ans])
    hm[array[i]] = i
```

Now the time complexity is O(n)
