## 217. Contains Duplicates

for this we can have multiple solutions one  of which is brute force in which we will check every number by matching by every number in the array.

```python
n = [1, 3, 2, 1]
found = False

if not found:
    for i in range(len(n)):
        for j in range(i, (len(n))):
            if i != j:
                if n[i] == n[j]:
                    print(True)
                    found = True
                    break

if not found:
    print(False)

```

This proposed solution had a time complexity of O(n*2) The question is if i can solve this by reducing the time complexity to O(n)

```python

n = [1,2,3,1]

spn = set()
found = False

if not found:
    for i in n: 
        if i not in spn:
            spn.add(i)
        else:
            print(True)
            found = True
            break

if not found:
    print(False)


```

## 242. Valid Anagram 

We have to find that the given strings are anagram of each other or not if they are then return True and if they are not return False.

***An Anagram is a word or string formed by rearranging the characters of another word or string, using every character exactly the same number of times.***

### Attempt 1

```python
s = "anagrams"
t = "nagramas" 

flag = True

if flag == True:
    for i in s: 
        if i not in t:
            print(False)
            flag = False
            break

if flag: 
    print(True)


```

This code will fail if given s = aab and t = abb the code will give true but the correct answer is false because the number should match as well.

### Attempt 3 with some help of neetcode

```python
s = "asdf"
t= "fdsa"

flag = True

if len(s) == len(t):

    countS, countT= {}, {}

    for i in range(len(s)):
        countS[s[i]] = 1 + countS.get(s[i], 0)
        countT[t[i]] = 1 + countT.get(t[i], 0)

    for c in countS:
        if countS[c] != countT.get(c):
            print(False)
            flag = False
            break

else:
    print(False)

if flag:
    print(True)
```

This is the optimal solution and this give O(logn)
